## TASK 1

```python
import os
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# Create directory if it does not exist
os.makedirs('images', exist_ok=True)

# Define paths (replace filenames with your own if needed)
path_img1 = os.path.join('images', 'cameraman.png')
path_img2 = os.path.join('images', 'cruyff.png')

# Fallback: create sample test patterns if images are missing
if not os.path.exists(path_img1):
    print("Notice: 'cameraman.tif' not found. Creating a synthetic image...")
    sample1 = (np.indices((256, 256))[0] ^ np.indices((256, 256))[1]).astype(np.uint8)
    cv2.imwrite(path_img1, sample1)

if not os.path.exists(path_img2):
    print("Notice: 'lena_gray_256.tif' not found. Creating a synthetic image...")
    sample2 = (np.indices((256, 256))[0] * 0.5 + np.indices((256, 256))[1] * 0.5).astype(np.uint8)
    Image.fromarray(sample2).save(path_img2)

print("Setup completed. All dependencies and files are ready.")
```

---

## TASK 2

### a

```python
# Task #2: a. Loading/reading an image using OpenCV
img = cv2.imread(path_img1)

# OpenCV loads color as BGR; convert if multi-channel, else render as grayscale
if len(img.shape) == 3 and img.shape[2] == 3:
    img_display = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    color_map = None
else:
    img_display = img
    color_map = 'gray'

plt.figure(figsize=(4, 4))
plt.imshow(img_display, cmap=color_map)
plt.title("Loaded with OpenCV")
plt.axis('off')
plt.show()
```

### b

```python
# Task #2: b. Loading/reading image using PIL
img2 = Image.open(path_img2)

plt.figure(figsize=(4, 4))
plt.imshow(img2, cmap=cm.Greys_r)
plt.title("Loaded with PIL")
plt.axis('off')
plt.show()
```

## Assessment - Array Operations with NumPy

```python
# Assessment: Operations on the array using NumPy
print("Array Data Type:", img_array.dtype)
print("Minimum Pixel Value:", img_array.min())
print("Maximum Pixel Value:", img_array.max())
print("Mean Intensity Value:", f"{img_array.mean():.2f}")

# 1. Negative / Invert Transformation
inverted_img = 255 - img_array

# 2. Slicing / Center Crop
h, w = img_array.shape[:2]
cropped_img = img_array[h//4 : 3*h//4, w//4 : 3*w//4]

# 3. Binary Thresholding (> 128)
binary_img = np.where(img_array > 128, 255, 0).astype(np.uint8)

# Visualization
fig, axes = plt.subplots(1, 4, figsize=(14, 4))

axes[0].imshow(img_array, cmap='gray')
axes[0].set_title("Original")

axes[1].imshow(inverted_img, cmap='gray')
axes[1].set_title("Inverted (Negative)")

axes[2].imshow(cropped_img, cmap='gray')
axes[2].set_title("Cropped")

axes[3].imshow(binary_img, cmap='gray')
axes[3].set_title("Thresholded (>128)")

for ax in axes:
    ax.axis('off')

plt.tight_layout()
plt.show()
```
